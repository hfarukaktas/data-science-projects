## Data Cleaning and Preprocessing

```python
# filling the null data on rain column with 0
weather_df["rain"] = weather_df["rain"].fillna(0)
```

```python
# transforming timestamps to the datetime variable type first and then to the date type.
# warning: the values on time_stamp column on cab_rides_df are in milliseconds, but on the weather_df they are in seconds.
cab_rides_df["date"] = pd.to_datetime(cab_rides_df["time_stamp"], unit = "ms").dt.date
weather_df["date"] = pd.to_datetime(weather_df["time_stamp"], unit = "s").dt.date

weather_df.head()
```

```python
# let's calculate the mean values for same day and location records in weather_df
weather_grouped = weather_df.drop(columns = "time_stamp").groupby(["location", "date"]).mean(numeric_only=True).reset_index()
weather_grouped.head()
```

```python
# merge the dataframes on location and date columns
merged_df = cab_rides_df.merge(weather_grouped, left_on = ["source", "date"], right_on = ["location", "date"], how = "left")

merged_df.head()
```

```python
# checking for cab rides that don't have matching weather info
merged_df.info()
```

```python
# dropping the rows with null values in the target variable "price"
df_clean = merged_df.dropna(subset = ["price"]).copy()
```

```python
df_clean.head()
```

## Explanatory Data Analysis

```python
# correlation heatmap
numeric_cols = df_clean.select_dtypes(include = ["float64", "int64"]).columns
corr_matrix = df_clean[numeric_cols].corr()

fig, ax = plt.subplots(figsize = (8,6))
sns.heatmap(corr_matrix, fmt = ".2f",annot = True, cbar = True, ax=ax)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

By looking at this correlation heatmap, we can say weather conditions didn't make much of a change on cab prices.

```python
# price distribution by cab type and service provider
price_by_cab = df_clean.groupby("cab_type")["price"].mean().round(2).reset_index()
price_by_cab_type = df_clean.groupby("product_id")["price"].mean().round(2).reset_index()

fig, axes = plt.subplots(1,2, figsize = (14,8))
sns.barplot(price_by_cab, x = "cab_type", y = "price", ax=axes[0])
axes[0].set_title("Price by Service Provider")
axes[0].set_xlabel("Brands")
axes[0].set_ylabel("Price")
sns.barplot(price_by_cab_type, x = "product_id", y = "price", hue = "product_id", palette="pastel", ax=axes[1])
axes[1].set_title("Price by Cab Type")
axes[1].set_xlabel("Ride Plans")
axes[1].set_ylabel("Price")
plt.xticks(rotation = 45)
plt.tight_layout()
plt.show()
```

There is some unhandled data on product_id column, so we are not able to understand which ride plans Uber offers to their customers.

```python
# price distribution by car segment
price_by_name = df_clean.groupby("name")["price"].mean().round(2).reset_index().sort_values(by = "price", ascending=False)

fig = plt.figure(figsize = (8,6))
sns.barplot(price_by_name, x = "name", y = "price", hue = "name", palette = "flare")
plt.title("Price by Car Segment")
plt.xlabel("Car Segment")
plt.ylabel("Price")
plt.xticks(rotation = 45)
plt.tight_layout()
plt.show()
```

```python
# price distrubution vs normal distribution
from scipy.stats import norm, lognorm

plt.figure(figsize = (12,6))
sns.set_theme(style="whitegrid")
sns.histplot(df_clean, x = "price", kde = True, stat = "density",bins = 80, label = "Real Distrubution")

# plotting the normal distribution line
mu, std = df_clean["price"].mean(), df_clean["price"].std()
x = np.linspace(df_clean["price"].min(), df_clean["price"].max(), 100)
# probability density function
p = norm.pdf(x, mu, std)
plt.plot(x, p, "r",linestyle = "--", linewidth = 2, label = f"Theoretical Normal (mu = {mu: .2f}, std = {std: .2f})")

shape, loc, scale = lognorm.fit(df_clean["price"], floc = 0)
# probability density function
p_lognorm = lognorm.pdf(x, shape, loc = loc, scale = scale)
plt.plot(x, p_lognorm, color = "forestgreen", linestyle = "-", linewidth = 3, label = "Theoretical Log-Normal Distribution")

plt.title("Cab Ride Price Distribution")
plt.xlabel("Price")
plt.ylabel("Density")
plt.legend()
plt.grid(axis = "y", alpha = 0.3)
plt.show()
```

Log-Normal distribution line looks like the best fit for our price distribution.

```python
# let's see the mean prices for each surge multiplier and outliers.
plt.figure(figsize = (12,8))
sns.boxplot(df_clean, x = "surge_multiplier", y = "price", hue = "cab_type", palette = "Set2")
plt.title("Prices by Surge Multiplier and Brand")
plt.xlabel("Surge Multiplier")
plt.ylabel("Price")
plt.legend()

plt.show()
```

This graph show us that Uber actually never using that surge multiplier, but still they have too much outlier price tags.
